```python
import pandas as pd
import pymc as pm
import arviz as az
import numpy as np
```

```python
def model_info(pymc_model_df):
    
    y = pymc_model_df['Normalised_Pole_Gap'].to_numpy(dtype = 'float64')

    drivers = pd.Index(sorted(pymc_model_df['Driver'].unique()), name = 'driver')
    constructors = pd.Index(sorted(pymc_model_df['Team'].unique()), name = 'constructor')
    rounds = pd.Index(sorted(pymc_model_df['Season_Round'].unique()), name = 'round')

    driver_no_id = pymc_model_df['Driver'].map({d:i for i, d in enumerate(drivers)}).to_numpy()
    constructor_no_id = pymc_model_df['Team'].map({c:i for i, c in enumerate(constructors)}).to_numpy()
    round_no_id = pymc_model_df['Season_Round'].map({r:i for i, r in enumerate(rounds)}).to_numpy()

    n_drivers = len(drivers)
    n_constructors = len(constructors)
    n_rounds = len(rounds)

    with pm.Model() as model:
        driver = pm.HalfNormal('driver', 0.01)
        constructor = pm.HalfNormal('constructor', 0.02)
        round = pm.HalfNormal('round', 0.005)
        # how much the variataion is from the 'average' - priors based on prior academic lit results
    
        raw_driver = pm.Normal('raw_driver', 0, 1, shape = n_drivers)
        raw_constructor = pm.Normal('raw_constructor', 0, 1, shape = n_constructors)
        raw_round = pm.Normal('raw_round', 0, 1, shape = n_rounds)

        driver_effect = pm.Deterministic('driver_effect', raw_driver * driver)
        constructor_effect = pm.Deterministic('constructor_effect', raw_constructor * constructor)
        round_effect = pm.Deterministic('round_effect', raw_round * round)

        sigma = pm.HalfNormal('sigma', 0.01)

        mean_pole_gap = driver_effect[driver_no_id] + constructor_effect[constructor_no_id] + round_effect[round_no_id]

        y_observations = pm.Normal('y_observations', mu = mean_pole_gap, sigma = sigma, observed = y)

        pymc_results = pm.sample(1000, tune = 1000, chains = 4, target_accept = 0.9, random_seed = 24)
        
        return pymc_results
```

```python
pymc_quali_2018_results = model_info(pd.read_csv('2018_season_quali_dataset.csv'))
```

```python
pymc_quali_2019_results = model_info(pd.read_csv('2019_season_quali_dataset.csv'))
```

```python
pymc_quali_2024_results = model_info(pd.read_csv('2024_season_quali_dataset.csv'))
```

```python
pymc_quali_2025_results = model_info(pd.read_csv('2025_season_quali_dataset.csv'))
```

```python
summaries = []

summaries.append(az.summary(pymc_quali_2018_results, var_names=['driver','constructor','round','sigma'], round_to=5).assign(season=2018))
summaries.append(az.summary(pymc_quali_2019_results, var_names=['driver','constructor','round','sigma'], round_to=5).assign(season=2019))
summaries.append(az.summary(pymc_quali_2020_results, var_names=['driver','constructor','round','sigma'], round_to=5).assign(season=2020))
summaries.append(az.summary(pymc_quali_2021_results, var_names=['driver','constructor','round','sigma'], round_to=5).assign(season=2021))
summaries.append(az.summary(pymc_quali_2022_results, var_names=['driver','constructor','round','sigma'], round_to=5).assign(season=2022))
summaries.append(az.summary(pymc_quali_2023_results, var_names=['driver','constructor','round','sigma'], round_to=5).assign(season=2023))
summaries.append(az.summary(pymc_quali_2024_results, var_names=['driver','constructor','round','sigma'], round_to=5).assign(season=2024))
summaries.append(az.summary(pymc_quali_2025_results, var_names=['driver','constructor','round','sigma'], round_to=5).assign(season=2025))

all_summaries = pd.concat(summaries)

all_summaries.to_csv('quali_pymc_summary.csv')
```
